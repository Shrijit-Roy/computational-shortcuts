```python
a = int(input())
b = []
for i in range(1, a + 1, 1):
    if a % i == 0:
        b.append(i)
print(b)
j = 0
k = len(b) - 1
while j <= len(b) / 2 and k >= len(b) / 2:
    c = b[j]
    d = b[k]
    print(c, "X", d)
    j = j + 1
    k = k - 1
```
